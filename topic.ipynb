**Question:** How has total rainfall in Washington changed from Jan 1, 2023 to Oct 27, 2026?

**Data Source:** https://nadp.slh.wisc.edu/networks/nadp-precipitation-network/

**Date Accessed:** 2026-11-07

```python
import pandas as pd
import os

directory = "data/"
target_prefix = "PRECIP_WA"
ignore_prefix = "SiteData"
total_rows = 0

for filename in os.listdir(directory):
    if filename.startswith(target_prefix) and not filename.startswith(ignore_prefix):
        file_path = os.path.join(directory, filename)
        data = pd.read_csv(file_path, header=1)
        cols = data.columns.tolist()
        site_id = filename.split("_")[1]
        total_rows += len(data)
        print(f"Site {site_id} has {len(data)} rows with columns as {cols}")
print("qualifiedPrecipInches: In")
print(f"There are a total of {total_rows} data entries across all 8 sites.")
        
```
